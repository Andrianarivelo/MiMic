# visualization toolkit demo
~
### run this notebook to see a demonstration of plots you can use to quickly analyze recordings
~
#### what will happen:
#### necessary packages are loaded
#### create an instance of the visualization class _Viz_
#### run a few plot examples for varying metrics
~
#### requirements:
#### run this notebook from the repository's `examples/` directory
#### the maintained example outputs for visualization live under `audios/mouse_1_outputs` and `audios/mouse_2_outputs`


```python
import os
import sys
sys.path.insert(0, '..')

from vocalpy.modules.viz import Viz
```

```python
# -- First we load the output object from VocalPy for a number of recording
# -- group the recordings as desired

# -- path to vocalpy object for each group
group1 = [
    os.path.join('audios', 'mouse_1_outputs', 'recording_without_spectrograms.vocalpy'),
    os.path.join('audios', 'mouse_1_outputs', 'recording_without_spectrograms.vocalpy'),
]

group2 = [
    os.path.join('audios', 'mouse_2_outputs', 'recording_without_spectrograms.vocalpy'),
    os.path.join('audios', 'mouse_2_outputs', 'recording_without_spectrograms.vocalpy'),
    os.path.join('audios', 'mouse_2_outputs', 'recording_without_spectrograms.vocalpy'),
]

group3 = [
    os.path.join('audios', 'mouse_1_outputs', 'recording_without_spectrograms.vocalpy'),
]

# -- set group names
group_names = [
    'isolation', 
    'reunion',
    'control',
]

# -- bin size in minutes
bin_size   = 1

# -- this will create a group visualization object called VizObj
VizObj = Viz(list_of_groups=[group1, group2, group3], 
                  group_names=group_names,
                  bin_size=bin_size)
print(VizObj)
```

# event plot for whole groups

```python
VizObj.plot(plot_type='group', dataname='raster')
```

# event plot for each recording

```python
VizObj.plot(plot_type='individual', dataname='raster')
```

# grouped plot for some metric
#### available metrics are: raster, rate, min_freq, max_freq, avg_freq, duration, bandwidth, min_intensity, max_intensity, avg_intensity, area.

```python
VizObj.plot(plot_type='group', dataname='avg_freq')
```

# individual recording plots for some metric
#### available metrics are: raster, rate, min_freq, max_freq, avg_freq, duration, bandwidth, min_intensity, max_intensity, avg_intensity, area.

```python
VizObj.plot(plot_type='individual', dataname='duration')
```
