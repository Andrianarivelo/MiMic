#### what will happen:

#### requirements:
#### run this notebook from the repository's `examples/` directory
#### `examples/audios/example_outputs/spectrogram` is a legacy image-only demo fixture; canonical CI baselines live under `mouse_1_outputs` and `mouse_2_outputs`


```python
import os
import sys
root_path = os.path.abspath(os.path.join('..'))
if root_path not in sys.path:
    sys.path.append(root_path)

import torch

import numpy as np
import matplotlib.pyplot as plt

from PIL import Image
from torch.autograd import Variable
from vocalpy.nn.classifier import VocalClassifier
```

```python
path_to_spectrogram = os.path.join('audios', 'example_outputs', 'spectrogram')
ClassClassifier = VocalClassifier(network_type='class', source=path_to_spectrogram, batch_size=1)
ClassClassifier.model.eval()
dataiter = iter(ClassClassifier.dataloader)
```

```python
for i in range(0,10):
    # -- first few examples are noise, for the noise CNN demo
    # -- skip them
    dataiter.next()

fig, axs = plt.subplots(2, 5, figsize=(20, 10))    
for ax in axs.reshape(-1):
    image = dataiter.next()
    image = image.to(ClassClassifier.device)
    image = Variable(image)
    score = ClassClassifier.model(image)
    _, predicted = torch.max(score.data, 1)
    ax.imshow(image.cpu().numpy()[0,0,:,:], cmap='gray')
    title = ClassClassifier.classes[predicted.numpy()[0]]
    ax.set_title(title, size=15)

fig.tight_layout()
plt.show()
```
